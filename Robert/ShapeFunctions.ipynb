```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rc
```

```python
plt.plot([[1],[0]], color="C0", label="\u03A6-1")
plt.plot(([0],[1]), color="C1", label="\u03A6-2")
plt.title("Linear shape functions")
plt.xlabel("x/L [-]")
plt.ylabel("u [-]")
plt.grid()
plt.legend()
plt.savefig("LinShapeFunc.png")
```

```python
x = np.linspace(0,1,101)
L = 1
yone = 1 -3*(x**2) + 2*(x**3)
ytwo = x - 2* (x**2/L**2) + x**3/L**3
ythree = 3*(x**2/L**2) -2*(x**3/L**3)
yfour = - x**2/L + x**3/L**2

plt.plot(x, yone, 'C3', label="\u03A6-3")
plt.plot(x, ytwo, "C4", label = "\u03A6-4")
plt.plot(x, ythree, "C5", label="\u03A6-5")
plt.plot(x, yfour, "C6", label="\u03A6-6")
plt.title("Hermite shape functions")
plt.xlabel("x/L [-]")
plt.ylabel("\u03BD [-] & \u03B8 [-]")
plt.grid()
plt.legend()
plt.savefig("hermiteshapefunc.png")
```
